```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng()

n_samples = 100000
prior_samples = rng.beta(1, 1, n_samples)

heads = 8
flips = 10
weights = (prior_samples**heads) * ((1 - prior_samples) ** (flips - heads))

posterior_samples = rng.choice(prior_samples, size=n_samples, p=weights / np.sum(weights))

plt.hist(prior_samples, bins=50, alpha=0.3, label="Prior ", density=True)
plt.hist(posterior_samples, bins=50, alpha=0.7, label="Posterior ", color="orange", density=True)
plt.axvline(0.8, color="red", linestyle="--", label="(0.8)")
plt.legend()
plt.show()
```
